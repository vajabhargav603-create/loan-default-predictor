### 3. EDA (20 Questions)

**Q1: What is the total number of records and features in the dataset ?**

```python
records, features = df.shape
print(f'Total Records: {records:,}')
print(f'Total Features: {features}')
```

**Q2: What are the data types of each feature ?**

```python
print('Feature Data Types:')
print(df.dtypes)
```

**Q3: How many null (missing) values are there in each column ?**

```python
print('Null Values per Column:')
print(df.isnull().sum())
```

**Q4: What is the proportion of loan defaults in the dataset ?**

```python
print("Count:", df['Default'].value_counts())
print("---------------------------------")
print("Percentage:", df['Default'].value_counts(normalize=True).mul(100).round(2))
```

**Q5: How many borrowers have a co-signer ?**

```python
print("Count:",df['HasCoSigner'].value_counts())
print("---------------------------------")
print("Percentage:",df['HasCoSigner'].value_counts(normalize=True).mul(100).round(2))
```

**Q6: What is the distribution of the target variable Default across borrowers with vs without a co-signer ?**

```python
cosigner_default = (df.groupby('HasCoSigner')['Default'].value_counts(normalize=True).unstack() * 100).round(2)

print('Default Rate by Co-Signer :') 
print(cosigner_default)
```

**Q7: What are the unique categories in LoanPurpose and their counts ?**

```python
print(df['LoanPurpose'].value_counts())
```

**Q8: What is the distribution of education levels among borrowers ?**

```python
count = df['Education'].value_counts()
percentage = (df['Education'].value_counts(normalize=True) * 100).round(2)
print("Count:",count)
print("--------------------------------")
print("Percentage:",percentage)
```

**Q9: What are the minimum, maximum, mean, and median values of the Age feature ?**

```python
values = df['Age'].describe()
print("Minimum Age:", values['min'])
print("Maximum Age:", values['max'])
print("Mean Age:", values['mean'])
print("Median Age:",  values['50%'])
```

**Q10: What is the average income of borrowers who defaulted compared to those who did not ?**

```python
avg_income = df.groupby('Default')['Income'].mean()
print('Average Income by Default Status:')
print(f"Non-Defaulted (0): ${avg_income[0]:,.2f}")
print(f"Defaulted (1):     ${avg_income[1]:,.2f}")
print(f"Difference:        ${(avg_income[0] - avg_income[1]):,.2f}")
```

**Q11: How many unique employment types are there in the dataset, and which one is the most common ?**

```python
emp_types = df['EmploymentType'].value_counts()
print(f'Number of unique employment types: {df["EmploymentType"].nunique()}')
print('Employment Type distribution:')
print(emp_types)
```

**Q12: What is the distribution of LoanTerm values ?**

```python
print(df['LoanTerm'].value_counts())
```

**Q13: Is there any clear correlation between CreditScore and InterestRate ?**

```python
correlation = df['CreditScore'].corr(df['InterestRate'])
print(f'correlation: {correlation:.4f}')
```

**Q14: What are the statistics (mean, median, std) of the LoanAmount feature ?**

```python
values = df['LoanAmount'].describe()
print(f"Mean Loan Amount:              ${values['mean']:,.2f}")
print(f"Median Loan Amount:            ${values['50%']:,.2f}")
print(f"Standard Deviation of Loan:    ${values['std']:,.2f}")
```

**Q15: What proportion of borrowers have a mortgage ?**

```python
print("Count:", df['HasMortgage'].value_counts())
print("---------------------------------")
print("Percentage:", df['HasMortgage'].value_counts(normalize=True).mul(100))
```

**Q16: How many borrowers have dependents ?**

```python
dep_counts = df['HasDependents'].value_counts()
print(dep_counts)
```

**Q17: What is the distribution of the number of credit lines ?**

```python
credit_lines = df['NumCreditLines'].value_counts().sort_index()
print(credit_lines)
```

**Q18: What is the average Debt-to-Income ratio for defaulted vs non-defaulted loans?**

```python
DTI_Avg = df.groupby('Default')['DTIRatio'].mean()
print('Average DTI Ratio by Default Status:')
print(f"Non-Defaulted (0): {DTI_Avg[0]:.4f}")
print(f"Defaulted (1):     {DTI_Avg[1]:.4f}")
print(f"Difference:        {DTI_Avg[1] - DTI_Avg[0]:.4f}")
```

**Q19: What is the distribution of employment length in months ?**

```python
months_stats = df['MonthsEmployed'].describe()
print(months_stats)
```

**Q20: Are there any duplicate records in the dataset?**

```python
dup_count = df.duplicated().sum()
print(f'Number of completely duplicate rows: {dup_count}')
id_dup = df['LoanID'].duplicated().sum()
print(f'Number of duplicate LoanID values: {id_dup}')
```
