### 1. Problem Statement & Background

**Problem Statement:**
Loan default prediction is a critical challenge in retail banking. When borrowers default on their loans, financial institutions suffer direct capital losses. Conversely, rejecting creditworthy borrowers limits interest revenue. The objective of this project is to analyze historical borrower data and build a classification model to predict whether a borrower will default on a loan (`Default = 1`) or not (`Default = 0`).

In Week 1, we focus on **Problem Definition and Dataset Exploration**. This involves understanding the structure of the data, inspecting column data types, identifying missing values, class distributions, and answering key analytical questions to formulate initial observations.

### 2. Environment Setup & Data Loading

```python
import pandas as pd
import numpy as np
import os

# Set display options to view all columns
pd.set_option('display.max_columns', None)
print('Pandas version:', pd.__version__)
print('NumPy version:', np.__version__)
```

```python
# Load the dataset
dataset_path = '../Loan_default.csv'
if not os.path.exists(dataset_path):
    dataset_path = 'd:/Darshan/Sem5/ML/Project/Loan_default.csv'

df = pd.read_csv(dataset_path)
print('Dataset loaded successfully!')
print('Dataset Shape (rows, columns):', df.shape)
```

### 3. Dataset Exploration (20 Questions)

**Q1: What is the total number of records and features in the dataset?**

```python
records, features = df.shape
print(f'Total Records (Rows): {records:,}')
print(f'Total Features (Columns): {features}')
```

**Q2: What are the data types of each feature?**

```python
print('Feature Data Types:')
print(df.dtypes)
```

**Q3: How many null (missing) values are there in each column?**

```python
print('Null Values per Column:')
print(df.isnull().sum())
```

**Q4: What is the proportion of loan defaults in the dataset (class balance/imbalance)?**

```python
default_counts = df['Default'].value_counts()
default_percentages = df['Default'].value_counts(normalize=True) * 100
for val, count, pct in zip(default_counts.index, default_counts.values, default_percentages.values):
    label = 'Default (1)' if val == 1 else 'Non-Default (0)'
    print(f'{label}: {count:,} ({pct:.2f}%)')
```

**Q5: How many borrowers have a co-signer (HasCoSigner)?**

```python
cosigner_counts = df['HasCoSigner'].value_counts()
cosigner_pct = df['HasCoSigner'].value_counts(normalize=True) * 100
for val, count, pct in zip(cosigner_counts.index, cosigner_counts.values, cosigner_pct.values):
    print(f'CoSigner={val}: {count:,} ({pct:.2f}%)')
```

**Q6: What is the distribution of the target variable Default across borrowers with vs without a co-signer?**

```python
cosigner_default = df.groupby('HasCoSigner')['Default'].value_counts(normalize=True).unstack() * 100
print('Default Rate by Co-Signer Status (%):')
print(cosigner_default)
```

**Q7: What are the unique categories in LoanPurpose and their counts?**

```python
purpose_counts = df['LoanPurpose'].value_counts()
purpose_pct = df['LoanPurpose'].value_counts(normalize=True) * 100
for val, count, pct in zip(purpose_counts.index, purpose_counts.values, purpose_pct.values):
    print(f'{val:12}: {count:,} ({pct:.2f}%)')
```

**Q8: What is the distribution of education levels among borrowers?**

```python
edu_counts = df['Education'].value_counts()
edu_pct = df['Education'].value_counts(normalize=True) * 100
for val, count, pct in zip(edu_counts.index, edu_counts.values, edu_pct.values):
    print(f'{val:12}: {count:,} ({pct:.2f}%)')
```

**Q9: What are the minimum, maximum, mean, and median values of the Age feature?**

```python
age_stats = df['Age'].describe()
print(f"Minimum Age: {age_stats['min']:.1f} years")
print(f"Maximum Age: {age_stats['max']:.1f} years")
print(f"Mean Age:    {age_stats['mean']:.2f} years")
print(f"Median Age:  {age_stats['50%']:.1f} years")
```

**Q10: What is the average income of borrowers who defaulted compared to those who did not?**

```python
avg_income = df.groupby('Default')['Income'].mean()
print('Average Income by Default Status:')
print(f"Non-Defaulted (0): ${avg_income[0]:,.2f}")
print(f"Defaulted (1):     ${avg_income[1]:,.2f}")
print(f"Difference:        ${(avg_income[0] - avg_income[1]):,.2f} lower for defaulted")
```

**Q11: How many unique employment types are there in the dataset, and which one is the most common?**

```python
emp_types = df['EmploymentType'].value_counts()
print(f'Number of unique employment types: {df["EmploymentType"].nunique()}')
print('Employment Type distribution:')
print(emp_types)
```

**Q12: What is the distribution of LoanTerm values?**

```python
term_counts = df['LoanTerm'].value_counts()
term_pct = df['LoanTerm'].value_counts(normalize=True) * 100
print('Loan Term Distribution (Months):')
for val, count, pct in zip(term_counts.index, term_counts.values, term_pct.values):
    print(f'{val} months: {count:,} ({pct:.2f}%)')
```

**Q13: Is there any clear correlation between CreditScore and InterestRate?**

```python
corr_coef = df['CreditScore'].corr(df['InterestRate'])
print(f'Pearson correlation coefficient between CreditScore and InterestRate: {corr_coef:.4f}')
```

**Q14: What are the statistics (mean, median, std) of the LoanAmount feature?**

```python
loan_stats = df['LoanAmount'].describe()
print(f"Mean Loan Amount:   ${loan_stats['mean']:,.2f}")
print(f"Median Loan Amount: ${loan_stats['50%']:,.2f}")
print(f"Std Dev of Loan:    ${loan_stats['std']:,.2f}")
```

**Q15: What proportion of borrowers have a mortgage (HasMortgage)?**

```python
mortgage_counts = df['HasMortgage'].value_counts()
mortgage_pct = df['HasMortgage'].value_counts(normalize=True) * 100
for val, count, pct in zip(mortgage_counts.index, mortgage_counts.values, mortgage_pct.values):
    print(f'Has Mortgage = {val}: {count:,} ({pct:.2f}%)')
```

**Q16: How many borrowers have dependents (HasDependents)?**

```python
dep_counts = df['HasDependents'].value_counts()
print('Borrowers with Dependents:')
print(dep_counts)
```

**Q17: What is the distribution of the number of credit lines (NumCreditLines)?**

```python
credit_lines = df['NumCreditLines'].value_counts().sort_index()
print('Number of Credit Lines Distribution:')
print(credit_lines)
```

**Q18: What is the average Debt-to-Income (DTI) ratio (DTIRatio) for defaulted vs non-defaulted loans?**

```python
avg_dti = df.groupby('Default')['DTIRatio'].mean()
print('Average DTI Ratio by Default Status:')
print(f"Non-Defaulted (0): {avg_dti[0]:.4f}")
print(f"Defaulted (1):     {avg_dti[1]:.4f}")
print(f"Difference:        {avg_dti[1] - avg_dti[0]:.4f} higher for defaulted")
```

**Q19: What is the distribution of employment length in months (MonthsEmployed)?**

```python
months_stats = df['MonthsEmployed'].describe()
print('Months Employed descriptive statistics:')
print(months_stats)
```

**Q20: Are there any duplicate records in the dataset?**

```python
dup_count = df.duplicated().sum()
print(f'Number of completely duplicate rows: {dup_count}')
id_dup = df['LoanID'].duplicated().sum()
print(f'Number of duplicate LoanID values: {id_dup}')
```

### 4. Summary of Week 1 Observations

Based on the questions answered above, we can draw the following initial observations:

1. **Imbalance Target:** The dataset is highly imbalanced with only **11.61%** defaults. This class imbalance must be accounted for during model evaluation (using metrics like F1-Score, Precision-Recall AUC instead of simple Accuracy) and possibly model training (class weights).
2. **Zero Missing Data & Duplicates:** The dataset is extremely clean, containing exactly zero missing values and zero duplicate rows or LoanIDs. This simplifies data cleaning.
3. **Co-Signer Impact:** Having a co-signer reduces default risk. Borrowers with co-signers have a default rate of **10.36%**, compared to **12.87%** for those without.
4. **Financial Differences:** Borrowers who default tend to have a significantly **lower average income** ($71,844 vs $83,899) and **higher average Debt-to-Income (DTI) ratio** (0.52 vs 0.50) compared to those who keep up with payments.
5. **Categorical Features:** Loan purpose, marital status, and education are evenly distributed across categories, providing balanced contextual dimensions for training classification algorithms.
